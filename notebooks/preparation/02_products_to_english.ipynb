```python
import pandas as pd
import numpy as np
```

```python
products = pd.read_csv("../data/raw/olist_products_dataset.csv")
products_english = pd.read_csv("../data/raw/product_category_name_translation.csv")
```

```python
products.shape
```

```python
category_dict = dict(
    zip(
        products_english["product_category_name"],
        products_english["product_category_name_english"]
    )
)
```

```python
category_dict
```

```python
products["product_category_name"] = products["product_category_name"].map(category_dict)
```

```python
products["product_category_name"] = products["product_category_name"].fillna("unknown")
```

```python
products["product_category_name"].isnull().sum()
```

```python
products["product_category_name"].unique()
```

```python
products.shape
```

```python
products.columns
```

```python
products.rename(columns={'product_name_lenght':'product_name_length',
                         'product_description_lenght':'product_description_length'
                         }, inplace=True)
```

```python
float_columns = [
    "product_name_length",
    "product_description_length",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

products[float_columns] = products[float_columns].astype("Int64")
```

```python
products.to_csv("../data/raw/olist_products_dataset_english.csv", index=False)
```
